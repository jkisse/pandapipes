```python
import pandapipes as ps
```

```python
net = ps.create_empty_network(name='net', fluid='lgas')

# nodes and branches
j1 = ps.create_junction(net, pn_bar=5.0, tfluid_k=283.15)
j2 = ps.create_junction(net, pn_bar=5.0, tfluid_k=283.15)
p1 = ps.create_pipe_from_parameters(net, from_junction=j1, to_junction=j2, length_km=10, diameter_m=0.3, k_mm=0.2)

# boundary conditions:
eg = ps.create_ext_grid(net, junction=j1, p_bar=5, t_k=283.15)
sink = ps.create_sink(net, junction=j2, mdot_kg_per_s=1.5)
```

```python
# calculation:
ps.pipeflow(net)
```

```python
# results:
print(net.res_junction)
```

```python
ps.plotting.simple_plot(net)
```

```python
# change fluid:
ps.create_fluid_from_lib(net, "hgas", overwrite=True)  # H-Gas, L-Gas, H2, water, ...
```
